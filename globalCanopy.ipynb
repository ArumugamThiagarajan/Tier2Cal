```python
import ee
ee.Authenticate()
ee.Initialize(project="ee-woodybiomass")
```

```python



# Step 3: Define your polygon (example coordinates)
polygon = ee.Geometry.Polygon([
    [[-109.760404,58.030997],
     [-110.1559118,49.0374031],
     [ -101.4107946,49.0085857],
     [-101.8063024,58.1239452]]
    ]
)

# Step 4: Define the image collection and filter it
image_collection = ee.ImageCollection("projects/meta-forest-monitoring-okw37/assets/CanopyHeight")\
                    .filterBounds(polygon)


```

```python
def download_image(image):
    image_id = image.id().getInfo()
    export_task = ee.batch.Export.image.toDrive(
        image=image,
        description=image_id,
        scale=10,
        region=polygon,
        fileFormat='GeoTIFF'
    )
    export_task.start()
    return image_id
```

```python
img_lst = image_collection.toList()
count = image_collection.getInfo()

for i in range(0, count):
    image = ee.Image(img_lst.get(i))
    name = image.get('system:index').getInfo()
    # print(name)
    task = ee.batch.Export.image(image, name)
    task.start()


```
